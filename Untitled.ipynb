```python
import sqlite3
import pandas as pd
import duckdb

DB_URI           = "file:./data/clean_database.db?mode=ro"
DASHBOARD_DB_URI = "file:./dashboard/data.sqlite"
```

```python
with sqlite3.connect(DB_URI, uri=True) as con:
    linkedin_df = pd.read_sql_query("""
        SELECT 
            (salary_min + salary_max) / 2 AS salary,
            searched_job_title AS job_title
            searched_country AS country,
            skills
        FROM linkedin
        """, con)
    
    upwork_df = pd.read_sql_query("""
        SELECT  hour_pay, skills, earnings
        FROM upwork
        """, con)

with sqlite3.connect(DASHBOARD_DB_URI, uri=True) as con:
    linkedin_df.to_sql("linkedin", con, if_exists="replace", index=False)
    upwork_df.to_sql("upwork", con, if_exists="replace", index=False)
```

```python
def format_for_json(chart_data: pd.DataFrame) -> dict | list:
    """
    Helper function that converts the output of a query into JSON ready dict
    so it can be used within the JS dashboard.
    params:
        chart_data is the output of the query that'll be used in the dashboard
        NOTE: it should feature a `country` column or `job_title` to be able to
        integrate with the filter feature in the dashboard
        
    returns: JSON ready dict (for our context) with the following format if 
        `country` and `job_title exist`:
        `{country_x: {job_title_y: [{x: 32, y: 21}, {x: 12, y: 43}, ..], ..}, ..}`
        and if it doesn't contain `country` and `job_title` it should look like:
        `[{x: 32, y: 21}, {x: 12, y: 43}, ...]`
    """
    records = chart_data.to_dict("records")
    formatted_data = {}
    
    for record in records:
        record_cpy = record.copy()
        country   = ""
        job_title = ""
        
        if "country" in record:
            country = record["country"]
            del record_cpy["country"]
            
        if "job_title" in record:
            job_title = record["job_title"]
            del record_cpy["job_title"]
        
        if country not in formatted_data:
            formatted_data[country] = dict()
    
        if job_title not in formatted_data[country]:
            formatted_data[country][job_title] = []
    
        formatted_data[country][job_title].append(record_cpy)

    return formatted_data
```

```python
avg_salary_data = duckdb.sql("""
    SELECT
        MEDIAN((salary_min + salary_max) / 2) AS salary,
        searched_country   AS country,
        searched_job_title AS job_title
    FROM linkedin_df
    WHERE salary_min IS NOT NULL AND
        pay_type = 'Annually' AND
        searched_country = 'United States'
    GROUP BY searched_job_title, searched_country
    HAVING COUNT(id) > 10
    ORDER BY salary DESC
""").df()
```

```python
mentioned_skills_data = duckdb.sql("""
    SELECT * FROM (
        SELECT 
            skill, job_title, country,
            SUM(occurance) AS occurances, 
            ROW_NUMBER() OVER (PARTITION BY job_title, country ORDER BY occurances DESC) AS index
        FROM (
            SELECT 
                UNNEST(SPLIT(skills, ',')) AS skill,
                searched_job_title AS job_title,
                searched_country AS country,
                1 AS occurance
            FROM linkedin_df
        )
        GROUP BY skill, job_title, country
    )
    WHERE index <= 5
    ORDER BY occurances DESC
""").df()
```

```python
avg_hour_pay_data = duckdb.sql("""
    SELECT AVG(hour_rate) AS avg_hour_pay,
        searched_job_title AS job_title
    FROM upwork_df
    GROUP BY searched_job_title
    ORDER BY avg_hour_pay
""").df()
```

```python
format_for_json(mentioned_skills_data)
```
